## 3. Draw Probability Calculator

Calculate exact probabilities of drawing specific cards from a deck.


```python
# Example 1: Probability of drawing at least one Pikachu (4 copies) in opening hand
result1 = pokemon_draw_probability_tool.invoke({
    "calculation_type": "at_least_n",
    "deck_size": 60,
    "draw_size": 7,
    "count_a": 4,
    "min_copies": 1
})
print("Example 1:", result1)
print()

# Example 2: Probability of drawing at least 2 Pikachu
result2 = pokemon_draw_probability_tool.invoke({
    "calculation_type": "at_least_n",
    "deck_size": 60,
    "draw_size": 7,
    "count_a": 4,
    "min_copies": 2
})
print("Example 2:", result2)
print()

# Example 3: Probability of drawing Pikachu OR Charmander
result3 = pokemon_draw_probability_tool.invoke({
    "calculation_type": "or",
    "deck_size": 60,
    "draw_size": 7,
    "count_a": 4,
    "count_b": 3
})
print("Example 3:", result3)
print()

# Example 4: Probability of drawing both Energy AND Supporter
result4 = pokemon_draw_probability_tool.invoke({
    "calculation_type": "and",
    "deck_size": 60,
    "draw_size": 7,
    "count_a": 10,
    "count_b": 8
})
print("Example 4:", result4)
```
